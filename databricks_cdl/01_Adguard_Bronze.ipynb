#Reading From JSON File

```python
bronze_df = (spark.read
    .format("json")
    .option("inferSchema", "true")
    .option("multiLine", "true") 
    .load('/Volumes/adguard_telemetry/bronze/source_system/raw_telemetry.json'))
display(bronze_df)
```

```python
bronze_df.printSchema()
```

#Writing to Bronze Layer

```python
bronze_df.write.mode("overwrite").saveAsTable("adguard_telemetry.bronze.adguard_raw")
```
