#Reading from Bronze Table

```python
df = spark.read.table("adguard_telemetry.bronze.adguard_raw")
```

#Data Tranformation

```python
from pyspark.sql.functions import col
import pyspark.sql.functions as F
df = (df.withColumn("query_domain",col("question.name"))
.withColumn("query_type",col("question.type"))
.withColumn("client_name",col("client_info.name"))
.withColumn("elapsed_ms",col("elapsedMs").cast("double"))
.select(
    "client",
    "client_name",
    "query_domain",
    "query_type",
    "reason",
    "elapsed_ms",
    "cached",
    F.col("time").cast("timestamp").alias("query_time")
))
display(df)
```

#Writing to Silver Layer

```python
(df.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true") 
    .saveAsTable("adguard_telemetry.silver.adguard_clean")
)
```
